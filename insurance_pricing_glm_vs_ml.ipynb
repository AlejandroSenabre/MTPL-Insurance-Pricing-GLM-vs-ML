```python
# 1. Merge as we discussed before
sev_agg = df_sev.groupby('IDpol')['ClaimAmount'].sum().reset_index()
df_final = pd.merge(df_freq, sev_agg, on='IDpol', how='left')
df_final['ClaimAmount'] = df_final['ClaimAmount'].fillna(0)

# 2. Cap the claims (Winsorization)
# We limit the claim amount to the 99th percentile to prevent extreme outliers
# from breaking the GLM model
cap = df_final['ClaimAmount'].quantile(0.99)
df_final['ClaimAmount_Capped'] = df_final['ClaimAmount'].clip(upper=cap)

# 3. Create the 'Pure Premium' target variable
# Pure Premium = Total Claim Cost / Exposure (This is your target for pricing)
df_final['PurePremium'] = df_final['ClaimAmount_Capped'] / df_final['Exposure']

print(df_final[['IDpol', 'ClaimNb', 'Exposure', 'PurePremium']].head())
```

```python
# Create a 'Claim Frequency' column (Number of claims / Exposure)
df_final['ClaimFreq'] = df_final['ClaimNb'] / df_final['Exposure']

# Create an 'Average Severity' column (Total cost / Number of claims, for those who had claims)
df_final['AvgSeverity'] = df_final['ClaimAmount_Capped'] / df_final['ClaimNb']
df_final['AvgSeverity'] = df_final['AvgSeverity'].fillna(0)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Check if higher BonusMalus (meaning worse history) leads to more claims
sns.boxplot(x='ClaimNb', y='BonusMalus', data=df_final[df_final['ClaimNb'] < 5])
plt.title('Relationship between BonusMalus and Number of Claims')
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. Let's select our core feature columns and the targets
# We include the variables described in the CAS documentation
feature_cols = ['VehPower', 'VehAge', 'DrivAge', 'BonusMalus', 'VehBrand', 'VehGas', 'Area', 'Density', 'Region']
target_cols = ['ClaimNb', 'ClaimAmount', 'Exposure']

# Filter data to keep things clean
df_modeling = df_final[feature_cols + target_cols].copy()

# 2. One-Hot Encoding for Categorical variables (VehBrand, VehGas, Area, Region)
# This converts text categories into columns of 0s and 1s
df_encoded = pd.get_dummies(df_modeling, columns=['VehBrand', 'VehGas', 'Area', 'Region'], drop_first=True)

# 3. Split the data into Train (80%) and Test (20%) sets
# We set a random_state so your results are reproducible
train_df, test_df = train_test_split(df_encoded, test_size=0.2, random_state=42)

print(f"Training set rows: {train_df.shape[0]}")
print(f"Testing set rows: {test_df.shape[0]}")
print("\nReady for modeling! Here are a few processed columns:")
display(train_df.head(2))
```

```python
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
import pandas as pd

# 1. Preparar las variables explicativas (X) y la variable objetivo (y)
# Para el GLM tradicional, seleccionaremos algunas variables clave para no saturar el modelo inicial
# NOTA: Al usar get_dummies antes, los nombres cambian (ej. VehGas_Regular, Area_B, etc.)

# Vamos a crear una lista con las variables numéricas y algunas de las dummies generadas
# Para simplificar el ejemplo, buscaremos columnas que existan en tu train_df
all_columns = train_df.columns.tolist()

# Filtramos las columnas que usaremos como predictores (excluyendo los targets y el IDpol si estuviera)
features_to_use = [col for col in all_columns if col not in ['ClaimNb', 'ClaimAmount', 'Exposure', 'IDpol', 'ClaimAmount_Capped', 'PurePremium', 'ClaimFreq', 'AvgSeverity']]

X_train = train_df[features_to_use].copy() # Explicitly create a copy to avoid SettingWithCopyWarning
y_train = train_df['ClaimNb']

# --- Start of Proposed Fix ---
# Ensure all columns in X_train are numeric.

# Convert boolean columns to int (0 or 1), as statsmodels might misinterpret them as objects
for col in X_train.select_dtypes(include='bool').columns:
    X_train[col] = X_train[col].astype(int)

# Identify any remaining non-numeric (object) columns
non_numeric_cols_X_train = X_train.select_dtypes(include='object').columns

if not non_numeric_cols_X_train.empty:
    print(f"Warning: Converting non-numeric columns in X_train to numeric: {list(non_numeric_cols_X_train)}")
    # Attempt to convert to numeric, coercing errors to NaN
    for col in non_numeric_cols_X_train:
        X_train[col] = pd.to_numeric(X_train[col], errors='coerce')
    # Fill NaNs with 0 after conversion, as these were likely dummy variables that might have errors
    X_train = X_train.fillna(0)
# --- End of Proposed Fix ---

# Es indispensable agregar una constante (intercepto) para el GLM en statsmodels
X_train_clean = sm.add_constant(X_train)

# 2. Definir el Offset (Exposición)
# El offset en una regresión Poisson DEBE ser el logaritmo de la exposición
offset_train = np.log(train_df['Exposure'])

# 3. Ajustar el Modelo Lineal Generalizado (GLM) con familia Poisson
print("Entrenando el Modelo de Frecuencia GLM (Poisson)...")
glm_poisson = sm.GLM(
    y_train,
    X_train_clean,
    family=sm.families.Poisson(),
    offset=offset_train
).fit()

# 4. Mostrar el resumen estadístico del modelo
print("\n=== RESUMEN DEL MODELO DE FRECUENCIA ===")
print(glm_poisson.summary())
```
